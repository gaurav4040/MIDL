```
import nibabel as nib
import matplotlib.pyplot as plt
prob = nib.load("pred_prob.nii.gz").get_fdata()
mask = nib.load("pred_mask.nii.gz").get_fdata()
slice_idx = prob.shape[0]//2
plt.figure(figsize=(10,5))
plt.subplot(1,2,1); plt.imshow(prob[slice_idx,:,:], cmap='gray'); plt.title('prob')
plt.subplot(1,2,2); plt.imshow(mask[slice_idx,:,:], cmap='gray'); plt.title('mask')
plt.show()
```
